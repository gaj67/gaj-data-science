## Laws of Expectations

Apart from the distributions themselves, a great part of probability theory is concerned with expectations of various functions of the stochastic variable. We present some useful rules and laws of expectations here, as well as some curiosities. Note that although we frequently specify $\mathbb{E}[X]$ and $\mathbb{V}[X]$ as the expectation and variance of $X$, respectively, it should always be kept in mind that these actually denote the expectation and variance of the
distribution $p(X)$ of $X$. Furthermore, $\mathbb{E}[f(X)]$ is the expectation of the quantity $f(X)$ with respect to the distribution of $X$, not the distribution of $f(X)$. We could more properly write this as either $\mathbb{E}_X[f(X)]$ 
or $\mathbb{E}_{p(X)}[f(X)]$ to avoid ambiguity — both forms are prevalent in the literature, along with other variants
such as $\mathbb{E}_p[f(X)]$, et cetera.
However, usually the correct context is clear enough to drop explicit reference to the stochastic variable and/or
its distribution.

### Expectation

Given some continuous stochastic variable $X\in\mathcal{X}$, we define the expectation of $f(X)$ with respect to the distribution $p(X)$ as
\begin{eqnarray}
\mathbb{E}[f(X)] & ~\doteq~ & \int_{\mathcal{X}} f(x)\,p(x)\,dx\,.
\end{eqnarray}
For a discrete variable, the corresponding expectation is defined as
\begin{eqnarray}
\mathbb{E}[f(X)] & ~\doteq~ & \sum_{x\in\mathcal{X}} f(x)\,p(x)\,dx\,,
\end{eqnarray}
instead.
Note that the
[law of the unconcious statistician](https://en.wikipedia.org/wiki/Law_of_the_unconscious_statistician "Wikipedia: Law of the unconcious statistician") (LOTUS) was coined, allegedly by Sheldon Ross, and perhaps humorously, to object to taking this definition axiomatically. Instead, it is claimed, the expectation should be derived theoretically anew for every distribution. However, some statisticians were not amused, and many statisticians apparently accept this definition without question, hence the "law", perhaps.

In practice, we do not expect to encounter any pathological distributions for which the definition fails. Note, however, that we **are** assuming that such expectations exist and are finite — the [Cauchy](https://en.wikipedia.org/wiki/Cauchy_distribution "Wikipedia: Cauchy distribution") distribution is one such exception for which the moments do not exist. 

### Iterated expectations

Given two stochastic variables, $X$ and $Y$, the *unconditional* expectation $\mathbb{E}[X]$ can be defined in terms of the
*conditional* expectation $\mathbb{E}[X\mid Y]$ via
\begin{eqnarray}
\mathbb{E}[X] & ~\doteq~ & \mathbb{E}[\,\mathbb{E}[X\mid Y]\,]\,.
\end{eqnarray}
This is sometimes referred to as the [law of iterated expectations](https://en.wikipedia.org/wiki/Law_of_total_expectation  "Wikipedia: Law of total expectation").
Note that the inner expectation is taken with respect to $X$ (or, actually, the conditional distribution $p(X\mid Y)$) for fixed $Y$,
whereas the outer expectation is taken with respect to variable $Y$ (or $p(Y)$).

As an example, suppose that a sample of values of $X$ may be partitoned into distinct groups based upon the corresponding values of $Y$. Then the law of iterated expectations indicates, informally, that the mean of the sample is the weighted mean of the group means, where the weightings take account of the numbers of samples in the various groups. 

### Variance

The variance, or spread, of the distribution $p(X)$ is defined as
\begin{eqnarray}
\mathbb{V}[X] & ~\doteq~ & \mathbb{E}\left[\left(X-\mathbb{E}[X]\right)^2\right]
\\ & ~=~ &
\mathbb{E}\left[X^2-2X\,\mathbb{E}[X]+\mathbb{E}[X]^2\right]
\\& ~=~ &
\mathbb{E}[X^2]-\mathbb{E}[X]^2\,.
\end{eqnarray}
Alternatively, the rearranged definition, namely
\begin{eqnarray}
\mathbb{E}[X^2] & ~\doteq~ & \mathbb{V}[X] + \mathbb{E}[X]^2\,,
\end{eqnarray}
is often more convenient when computing  moments.

### Iterated variances

Given two stochastic variables, $X$ and $Y$, the *unconditional* variance $\mathbb{V}[X]$ can be defined in terms of the
*conditional* expectation $\mathbb{E}[X\mid Y]$ and variance $\mathbb{V}[X\mid Y]$ via
\begin{eqnarray}
\mathbb{V}[X] & ~\doteq~ & \mathbb{E}[\,\mathbb{V}[X\mid Y]\,] 
+ \mathbb{V}[\,\mathbb{E}[X\mid Y]\,]
\,.
\end{eqnarray}
This is sometimes referred to as the [law of iterated variances](https://en.wikipedia.org/wiki/Law_of_total_variance  "Wikipedia: Law of total variance").
Gelman *et al.* [[GCSR1995]](references.ipynb#GCSR1995 "References: Bayesian Data Analysis") offer a short proof, given here in reverse:
\begin{eqnarray}
\mathbb{V}[X] & ~=~ & \mathbb{E}[X^2]-\mathbb{E}[X]^2
\nonumber\\& ~=~ &
\mathbb{E}[\,\mathbb{E}[X^2\mid Y]\,]
-\mathbb{E}[\,\mathbb{E}[X\mid Y]\,]^2
\nonumber\\& ~=~ &
\mathbb{E}\left[\,\mathbb{E}[X^2\mid Y]-\mathbb{E}[X\mid Y]^2\,\right]
+
\mathbb{E}[\,\mathbb{E}[X\mid Y]^2\,]-\mathbb{E}[\,\mathbb{E}[X\mid Y]\,]^2
\nonumber\\& ~=~ &
\mathbb{E}[\,\mathbb{V}[X\mid Y]\,] + \mathbb{V}[\,\mathbb{E}[X\mid Y]\,]
\,.
\end{eqnarray}

To continue with our [previous](#Iterated-expectations "Section: Iterated expectations") example, 
the first term on the right-hand side is the (weighted) mean of the group variances, and represents the
overall *within-group* variance. The second term on the right-hand side is the (weighted) variance of the group means, and represents the *between-group* variance.
Hence, the law of iterated variances informally states that the total variance is the sum of the within-group variance and the between-group variance. This is useful, for example, when measuring the compactness of clustering data points into groups. The within-group variance measures the dissimilarity between points within the same cluster, whereas the between-group variance measures the dissimilarity between the centroids of different clusters. A good clustering solution ahould attempt to minimise the former and maximise the latter.

### Covariance

Given two stochastic variables, $X$ and $Y$, the covariance between the distributions $p(X)$ and $p(Y)$
is given by
\begin{eqnarray}
\texttt{Cov}[X,Y] & ~\doteq~ & \mathbb{E}\left[\left(X-\mathbb{E}[X]\right)\,\left(Y-\mathbb{E}[Y]\right)\right]
\\ & ~=~ &
\mathbb{E}\left[\,XY-X\,\mathbb{E}[Y]-Y\,\mathbb{E}[X]+\mathbb{E}[X]\,\mathbb{E}[Y]\,\right]
\\& ~=~ & \mathbb{E}[XY]-\mathbb{E}[X]\,\mathbb{E}[Y]\,.
\end{eqnarray}
Noted that it [follows](#Variance "Section: Variance") that $\mathbb{V}[X]=\mathtt{Cov}[X,X]$.

### Law of iterated covariances

Given three stochastic variables, $X$, $Y$ and $Z$, the *unconditional* covariance $\mathtt{Cov}[X,Y]$ can be defined in terms of *conditional* expectations and covariance via
\begin{eqnarray}
\mathtt{Cov}[X,Y] & ~\doteq~ & \mathbb{E}[\,\mathtt{Cov}[X,Y\mid Z]\,] + 
\mathtt{Cov}[\,\mathbb{E}[X\mid Z],\mathbb{E}[Y\mid Z]\,]
\,.
\end{eqnarray}
This is sometimes referred to as the [law of iterated covariances](https://en.wikipedia.org/wiki/Law_of_total_covariance  "Wikipedia: Law of total covariance").


### Expectations of extremes

Given two stochastic variables, $X$ and $Y$, the *order statistics* that summarise the extreme values are given by
$X_{(1)}\doteq\min(X,Y)$ and $X_{(2)}\doteq\max(X,Y)$, such that $X_{(1)}\le X_{(2)}$. It can then be shown that
\begin{eqnarray}
 & X+Y & ~=~ \min(X,Y)+\max(X,Y)
\\
\Rightarrow~ & \mathbb{E}[X]+\mathbb{E}[Y] &~=~ \mathbb{E}[\min(X,Y)]+\mathbb{E}[\max(X,Y)]
\\
\Rightarrow~ & \mathbb{E}[\max(X,Y)] & ~=~ \mathbb{E}[X]+\mathbb{E}[Y] - \mathbb{E}[\min(X,Y)]\,.
\end{eqnarray}
This is sometimes written in the form
\begin{eqnarray}
\mathbb{E}[X\lor Y] & ~=~ \mathbb{E}[X]+\mathbb{E}[Y] - \mathbb{E}[X\land Y]\,,
\end{eqnarray}
where the operators '$\land$' and '$\lor$' are taken to represent minimisation and maximisation, respectively.
This form is chosen to correspond to the usual law of probabilities, i.e. either that
\begin{eqnarray}
p(X\cup Y) & ~=~ p(X)+p(Y) - p(X\cap Y)\,,
\end{eqnarray}
for sets $X$ and $Y$, where '$\cup$' and '$\cap$' represent set union and intersection, respectively, or else
\begin{eqnarray}
p(X\lor Y) & ~=~ p(X)+p(Y) - p(X\land Y)\,,
\end{eqnarray}
where $X$ and $Y$ are Boolean predicates, and '$\lor$' and '$\land$' represent logical disjunction and conjunction, 
respectively.
However, due to the ambiguity between operator definitions, I personally prefer the version in terms of explicit maximisation and minimisation.

### Mean field approximation

Given a stochastic variable $X\in\mathcal{X}$, consider a transformation 
$f:\mathcal{X}\mapsto\mathcal{Y}$, such that
$Y=f(X)\in\mathcal{Y}$.
Assuming that the function $f$ is sufficiently differentiable, then the Taylor series expansion of $f$ around
the mean $\mathbb{E}[X]\in\mathcal{X}$ is given by
\begin{eqnarray}
& f(X) & ~=~ f\left(\mathbb{E}[X]\right) 
+ f'\left(\mathbb{E}[X]\right)\,\left(X-\mathbb{E}[X]\right)
+\frac{1}{2}f''\left(\mathbb{E}[X]\right)\,\left(X-\mathbb{E}[X]\right)^2+\cdots
\\
\Rightarrow~ & \mathbb{E}[f(X)] & ~=~ f\left(\mathbb{E}[X]\right)
+ \frac{1}{2}f''\left(\mathbb{E}[X]\right)\,\mathbb{V}[X]+\cdots\,.
\end{eqnarray}
Provided that the terms on the right-hand side after the first term are negligable, then we
obtain the so-called *mean field approximation* (MFA):
\begin{eqnarray}
\mathbb{E}[f(X)] & ~\approx~ f\left(\mathbb{E}[X]\right)\,.
\end{eqnarray}
Note that we are using the word 'field' here in the sense of physics and not mathematics, e.g. an electric field
with a strength that varies in the spatial coordinates and time.
Informally, the MFA states that the mean of a function is approximately a function of the mean.

We can also approximate the variance of the tansformation via
\begin{eqnarray}
\mathbb{V}[f(X)] & ~=~ & \mathbb{E}\left[\left(f(X)-\mathbb{E}[f(X)]\right)^2\right]
\\& ~\approx~ & \mathbb{E}\left[f'(\mathbb{E}[X])^2\,\left(X-\mathbb{E}[X]\right)^2\right]
\\ & ~=~ & f'(\mathbb{E}[X])^2\,\mathbb{V}[X]\,,
\end{eqnarray}
provided that $f'(\mathbb{E}[X])\ne 0$. If this condition does not hold, then we can include an extra term
from the Taylor series expansion,
such that
\begin{eqnarray}
\mathbb{V}[f(X)] & ~\approx~ & \frac{1}{4}f''(\mathbb{E}[X])^2\,\mathbb{V}\left[\left(X-\mathbb{E}[X]\right)^2\right]\,,
\end{eqnarray}
instead, although this will require more work to evaluate.

## Exchangeability of Sequences

The famous contribution of de Finetti involves the notion of *exchangeability* and de Finetti's resulting *representation  theorem*.
A *finite* sequence $\mathbf{X}_N\doteq(X_1,\ldots,X_N)$ is said to be exchangeable if and only if every permutation of the sequence is *equi-probable*, i.e.
\begin{eqnarray}
p(X_1,\ldots,X_N) & ~=~ & p(X_{\pi_1},\ldots,X_{\pi_N})\,,
\end{eqnarray}
for  every permutation 
$\boldsymbol{\pi}\doteq(\pi_1,\ldots,\pi_N)\in\{1,\ldots,N\}^N$
with $\pi_i\ne\pi_j$ for $i\ne j$.

Note that exchangeability is a weaker condition than the assumption of *independence*.
For example, it can be shown [[J2002]](#references.ipynb#J2002 "Probability Theory. The Logic of Science") that a finite sequence of draws from an urn is exchangeable, despite the fact that each draw depends upon all previous draws.

It now follows that an *infinite* sequence is exchangeable if anly only if 
every finite sub-sequence is exchangeable.
Given an infinite exchangeable sequence, the representation theorem states
that there exists some latent parameter $\theta\in\Theta$ with distribution $p(\theta)$ such that
\begin{eqnarray}
p(X_1,\ldots,X_N) & ~=~ &
\int_\Theta \left\{\prod_{i=1}^{N}p(X_i\mid\theta)\right\}\,
p(\theta)\,d\theta\,.
\end{eqnarray}
In other words, the elements of an exchangeable sequence act as if they 
were conditionally independent, given $\theta$.
The representation theorem was heralded by Bayesians (such as de Finetti) as proof of the existence of prior distributions.
It is therefore ironic that, as we shall 
[see](#Binary-representation-theorem "Section: Binary representation theorem"), the proof of the theorem
depends heavily upon an *orthodox* statistical assumption,  namely
the existence of a limiting relative frequency in an infinite sequence of data.
Hence, the representation theorem could also be interpreted as a Bayesian  proof of *frequentist* probability!

### Binary exchangeable sequences

Consider an exchangeable binary sequence $\mathbf{X}_N\doteq(X_1,\ldots,X_N)$ where
$X_i\in\{0,1\}$. There are $2^N$ distinct binary sequences, but these are not, in general, all equi-probable. 
For convenience, we shall henceforth consider $X_i=1$ to represent a 'success' and $X_i=0$ to represent a failure.

Now define $S_N\doteq S(\mathbf{X}_N)=X_1+\cdots+X_N$ to be the number of successes in the sequence, such that $N-S_N$ is the number of failures. There are 
${N\choose s}$ distinct sequences having exactly $S_N=s$ successes,
for any $s=0,\ldots,N$. Due to exchangeability, all sequences restricted to $S_N=s$ are equi-probable permutations of each other, such that
\begin{eqnarray}
p(S_N=s) & ~=~ & {N\choose s}\,p(\mathbf{X}_N)\,,
\end{eqnarray}
for any single, given sequence $\mathbf{X}_N$ that has exactly $s$ successes. 
The representation theorem to be derived in the [following](#Binary-representation-theorem "Section: Binary representation theorem") section will
specify the form that $p(\mathbf{X}_N)$ must take.

Next, suppose we partition the sequence as
$\mathbf{X}_N=\mathbf{X}_n\oplus\mathbf{Y}_{N-n}$ for some $0\le n\le N$,
where $\mathbf{X}_n\doteq(X_1,\ldots,X_n)$
and $\mathbf{Y}_{N-n}\doteq(X_{n+1},\ldots,X_N)$.
By convention, to properly handle the edge cases of $n=0$ and $n=N$, we define $\mathbf{X}_0$ and $\mathbf{Y}_0$ as empty sequences
with a probability of unity.

Now suppose that the left sub-sequence $\mathbf{X}_n$ contains exactly $S_n=r$ successes, for some $r=0,\ldots,n$.
Since the sequence $\mathbf{X}_N$ is exchangeable, then so too are the sub-sequences
$\mathbf{X}_n$ and $\mathbf{Y}_{N-n}$. Hence, there are ${n\choose r}$
equi-probable left sub-sequences restricted to $S_n=r$ successes, and
${N-n\choose s-r}$ equi-probable right sub-sequences restricted to
the remaining $s-r$ successes.
Consequently, given $N$, $n$ and $S_N=s$, the probability that
$S_n=r$ is given by
\begin{eqnarray}
p(S_n=r\mid S_N=s) & ~=~ &
\frac{{n\choose r}\,{N-n\choose s-r}}{N\choose s}\,.
\end{eqnarray}
Note that, for combinatorial consistency, we define $p(S_n=r\mid S_N=s)=0$
if $s-r<0$ or $s-r>N-n$. The constraint that $s-r\ge 0$ corresponds to 
$S_n\le S_N$, meaning that a sub-sequence cannot have more successes than
its outer sequence. Likewise, the constraint that
$s-r\le N-n$ corresponds to $n-S_n\le N-S_N$, i.e.
a sub-sequence cannot have more failures than its outer sequence.

Conversely, if we treat $S_n=r$ as being fixed and $S_N=s$ as being variable, then
we obtain the lower bound $S_N\ge S_n$ and the upper bound $S_N\le N+S_n-n$.
It thus follows from the law of conditional probability that
\begin{eqnarray}
p(S_n=r) & ~=~ & \sum_{s=0}^{N}p(S_n=r\mid S_N=s)\,p(S_N=s)
\nonumber\\& ~=~ &
\sum_{s=r}^{N+r-n}p(S_n=r\mid S_N=s)\,p(S_N=s)\,,
\end{eqnarray}
using these bounds on $S_N$.

It turns out that $p(S_n\mid S_N)$ is just the
[hypergeometric](https://en.wikipedia.org/wiki/Hypergeometric_distribution "Wikipedia: Hypergeometric distribution") distribution. To see this, we
 expand the combinations into factorials and rearrange, namely
\begin{eqnarray}
p(S_n=r\mid S_N=s) & ~=~ &
\left.\frac{n!}{r!\,(n-r)!}\times\frac{(N-n)!}{(s-r)!\,(N-n-s+r)!}
\right/\frac{N!}{s!\,(N-s)!}
\nonumber\\& ~=~ &
\left.\frac{s!}{r!\,(s-r)!}\times\frac{(N-s)!}{(n-r)!\,(N-s-n+r)!}
\right/\frac{N!}{n!\,(N-n)!}
\nonumber\\& ~=~ &
\frac{{s\choose r}\,{N-s\choose n-r}}{N\choose n}
\,.
\end{eqnarray}
Alternatively, we can define the *falling factorial* as
\begin{eqnarray}
(m)_k & ~\doteq~ & \frac{m!}{(m-k)!}
~=~m\,(m-1)\,\cdots\,(m-k+1)\,,
\end{eqnarray}
which is the product of exactly $k$ terms. Consequently, a different
rearrangement of the above expansions then gives
\begin{eqnarray}
p(S_n=r\mid S_N=s) & ~=~ &
{n\choose r}\,\frac{(s)_r\,(N-s)_{n-r}}{(N)_n}\,.
\end{eqnarray}
This latter form will prove useful when deriving the asymptotic distribution
for $N\rightarrow\infty$, as shown in the
[following](#Binary-representation-theorem "Section: Binary representation theorem")  section.

### Binary representation theorem

We shall now derive the representation theorem for exchangeable binary sequences, which applies to the finite sequence $\mathbf{X}_N$. 
The first major assumption is that there exists some stochastic process that generates infinite, exchangeable, binary sequences, such that
$\mathbf{X}_N$ is in fact a sub-sequence of an underlying infinite sequence.
Note that the representation theorem does not hold true, in general, for purely finite sequence processes.
The second major assumption is that the sequence process possesses some constant, but unknown, parameter $\theta$ that governs the generation of successes.
In other words, $\theta$ represents the long-run frequency of successes.



For the finite sequence $\mathbf{X}_N$, we now define 
$\tilde{\theta}_N\doteq\frac{S_N}{N}$ to be the empirical proportion of successes, such that $S_N=s$ corresponds to $s=N\tilde{\theta}_N$,
for $s=0,1,\ldots,N$. 
Hence, from the [previous](#Binary-exchangeable-sequences "Section: Binary exchangeable sequences") section, we deduce that
\begin{eqnarray}
p(S_n=r\mid S_N=s) & ~=~ &
{n\choose r}\,\frac{(s)_r\,(N-s)_{n-r}}{(N)_n}
\nonumber\\& ~=~ &
{n\choose r}\,\frac{(N\tilde{\theta}_N)_r\,(N\,[1-\tilde{\theta}_N])_{n-r}}{(N)_n}
\,,
\end{eqnarray}
such that
\begin{eqnarray}
p(S_n=r) & ~=~ & \sum_{s=0}^{N}p(S_n=r\mid S_N=s)\,p(S_N=s)
\nonumber\\& ~=~ &
\sum_{\tilde{\theta}_N\in\tilde{\Theta}_N}
{n\choose r}\,\frac{(N\tilde{\theta}_N)_r\,(N\,[1-\tilde{\theta}_N])_{n-r}}{(N)_n}\,
p(\tilde{\theta}_N)\,,
\end{eqnarray}
where $\tilde{\Theta}_N\doteq\{\frac{0}{N},\frac{1}{N},\ldots,\frac{N}{N}\}$.

The equations above must hold true for any integer $N\ge n$, including as
$N\rightarrow\infty$, i.e. as $\mathbf{X}_N$ grows towards an infinite sequence as stipulated by our first assumption.
Note that, as $N\rightarrow\infty$, we must also have
$\tilde{\theta}_N\rightarrow\theta$, due to our second assumption.
Furthermore, we find that $\tilde{\Theta}_N\rightarrow[0,1]$ as $N\rightarrow\infty$, such that 
$p(\tilde{\theta}_N)\rightarrow p(\theta)d\theta$,
where the distribution $p(\theta)$ encodes our uncertainty about the true
value of $\theta$.
Consequently, we deduce that
\begin{eqnarray}
p(S_n=r) & ~=~ & {n\choose r}\,\lim_{N\rightarrow\infty}\int_0^1
\frac{(N\theta)_r\,(N\,[1-\theta])_{n-r}}{(N)_n}\,
p(\theta)\,d\theta\,.
\end{eqnarray}
Finally, under mild conditions, we assume that the order of the limit and the integral may be exchanged. We now recall from the
[definition](#Binary-exchangeable-sequences "Section: Binary exchangeable sequences") of a falling factorial that $(m)_k$ is the product of
exactly $k$ terms that decrease by unity at each step. Consequently, we
deduce that
\begin{eqnarray}
\lim_{N\rightarrow\infty}\frac{(N\theta)_r\,(N\,[1-\theta])_{n-r}}{(N)_n}
& ~=~ & 
\lim_{N\rightarrow\infty}\frac{(N\theta)^r\,[N(1-\theta)]^{n-r}}{N^n}
~=~\theta^r\,(1-\theta)^{n-r}\,,
\end{eqnarray}
such that
\begin{eqnarray}
p(S_n=r) & ~=~ & {n\choose r}\,\int_0^1 \theta^r\,(1-\theta)^{n-r}\,
p(\theta)\,d\theta\,.
\end{eqnarray}
We recognise this as a continuous mixture of binomial distributions.

In the limit as $N\rightarrow\infty$, the infinite sequence
$\mathbf{X}_\infty$ has a fixed success rate of $\theta$. Hence, due to exchangeability, any element $X_i$ of $\mathbf{X}_\infty$ is essentially
equivalent to any other element, such that
$p(X_i=1\mid\theta)=\theta$, for $i=1,2,\ldots,\infty$. It also follows 
from exchageability that
if the finite sub-sequence $\mathbf{X}_n$ contains exactly $r$ successes, then
the ${n\choose r}$ distinct permutations of $\mathbf{X}_n$ are
equi-probable, such that
\begin{eqnarray}
p(X_1,\ldots,X_n) & ~=~ & \int_0^1\left\{\prod_{i=1}^n p(X_i\mid\theta)\right\}
\,p(\theta)\,d\theta\,.
\end{eqnarray}
This is the representation theorem for finite, exchangeable, binary sequences
that are embedded within an infinite, exchangeable, binary sequence.
We recognise this last result as a continuous mixture of Bernoulli distributions. Note, however, that nowhere did we assume that the $X_i$
elements were independent, either conditionally or unconditionally.
Instead, we only assumed that they were unconditionally 
[exhangeable](#Binary-exchangeable-sequences "Section: Binary exchangeable sequences"),
which is a weaker requirement.
